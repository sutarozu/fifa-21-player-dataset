## Tahap 2: Pemeriksaan Kolom Value, Wage, dan Release Clause
Pemeriksaan ini bertujuan memahami cara penulisan nilai uang pada kolom Value, Wage, dan Release Clause, yaitu simbol mata uang, satuan K dan M, desimal, serta nilai nol, agar aturan konversinya ke angka dapat ditentukan dengan tepat pada tahap pembersihan. Tahap ini hanya memeriksa dan belum mengubah data.

### Bentuk Nilai dan Tipe Data
Kolom Value, Wage, dan Release Clause berisi teks yang diawali simbol € dan pada sebagian nilai diakhiri satuan M atau K, misalnya €75M dan €250K. Tipe datanya object dan tidak ada nilai kosong pada ketiga kolom, sehingga nilainya belum bisa dihitung sebagai angka.

```python
for kolom in ["Value", "Wage", "Release Clause"]:
    print(kolom, data[kolom].dtype, [repr(n) for n in data[kolom].head(10)])

print("\n", data[["Value", "Wage", "Release Clause"]].isna().sum(), sep="")
```

### Pola Unik
Setelah semua digit diganti dengan satu penanda, kolom Value memiliki 4 pola unik, kolom Wage memiliki 2 pola, dan kolom Release Clause memiliki 4 pola. Artinya formatnya tidak seragam dan konversi harus menangani setiap pola.

```python
for kolom in ["Value", "Wage", "Release Clause"]:
    pola = data[kolom].str.replace(r"\d+", "9", regex=True)
    print(kolom)
    print(pola.value_counts(), "\n")
```

### Akhiran
Kolom Value memiliki 3 kelompok nilai, yaitu 11.944 nilai berakhiran K, 6.786 nilai berakhiran M, dan 249 nilai tanpa akhiran. Kolom Wage memiliki 2 kelompok nilai, yaitu 14.824 nilai berakhiran K dan 4.155 nilai tanpa akhiran. Kolom Release Clause memiliki 3 kelompok nilai, yaitu 9.629 nilai berakhiran M, 8.088 nilai berakhiran K dan 1.262 nilai tanpa akhiran. Karena itu aturan konversi tidak boleh disamakan untuk semua kolom.

```python
for kolom in ["Value", "Wage", "Release Clause"]:
    print(kolom)
    print(data[kolom].str[-1:].value_counts(), "\n")
```

### Desimal dan Akhiran
Titik desimal hanya muncul pada nilai berakhiran M, yaitu 5.055 baris pada kolom Value dan 8.540 baris pada kolom Release Clause. Tidak semua nilai berakhiran M memiliki desimal, karena masih ada 1.731 baris pada Value dan 1.089 baris pada Release Clause yang berupa bilangan bulat. Nilai berakhiran K tidak pernah memiliki desimal, dan kolom Wage tidak memiliki titik desimal sama sekali. Karena itu desimal hanya perlu ditangani pada satuan M.

```python
for kolom in ["Value", "Wage", "Release Clause"]:
    ada_titik = data[kolom].str.contains(r"\.")
    akhiran = data[kolom].str[-1:]
    print(kolom)
    print(pd.crosstab(akhiran, ada_titik), "\n")
```

### Nilai Tanpa Akhiran
Pada kolom Value, seluruh 249 nilai tanpa akhiran bernilai €0, dan pada kolom Release Clause seluruh 1.262 nilai tanpa akhiran bernilai €0. Pada kolom Wage, nilai tanpa akhiran tidak selalu nol, karena dari 4.155 nilai tanpa akhiran hanya 238 yang bernilai €0. Sisanya, yaitu 3.917 baris, berupa nilai kecil seperti €500 sebanyak 2.413 baris, €900 sebanyak 184 baris, €600 sebanyak 183 baris, dan €850 sebanyak 178 baris. Nilai tanpa akhiran pada Wage terdiri dari 11 nilai unik, termasuk €0, dan seluruhnya sudah berupa euro penuh, sehingga tidak boleh dianggap nol dan tidak boleh dikalikan seperti satuan K.

```python
tanpa_akhiran = data["Value"][~data["Value"].str.endswith(("M", "K"))]
print(tanpa_akhiran.value_counts().head())
print(tanpa_akhiran.nunique())
```

```python
tanpa_akhiran = data["Wage"][~data["Wage"].str.endswith(("M", "K"))]
print(tanpa_akhiran.value_counts().head())
print(tanpa_akhiran.nunique())
```

```python
tanpa_akhiran = data["Release Clause"][~data["Release Clause"].str.endswith(("M", "K"))]
print(tanpa_akhiran.value_counts().head())
print(tanpa_akhiran.nunique())
```

### Nilai Nol
Nilai €0 berjumlah 249 pada kolom Value, 238 pada kolom Wage, dan 1.262 pada kolom Release Clause. Keputusan apakah nilai nol dianggap nol sebenarnya atau tidak diketahui diambil setelah membandingkan antar kolom.

```python
for kolom in ["Value", "Wage", "Release Clause"]:
    print(kolom, (data[kolom] == "€0").sum())
```

### Nol Antar Kolom
Sebanyak 238 pemain bernilai €0 pada kolom Value dan Wage sekaligus. Seluruh pemain dengan Wage nol juga memiliki Value nol, tetapi 11 pemain memiliki Value nol padahal Wage-nya bukan nol. Seluruh 249 pemain dengan Value nol juga memiliki Release Clause nol, sedangkan 1.013 pemain memiliki Release Clause nol padahal Value-nya bukan nol. Pola ini menunjukkan bahwa nol pada Release Clause tidak selalu sejalan dengan nol pada Value, sehingga keduanya perlu diperlakukan secara terpisah.

```python
nol_value = data["Value"] == "€0"
nol_wage = data["Wage"] == "€0"
nol_rel_clau = data["Release Clause"] == "€0"

print((nol_value & nol_wage).sum())
print((nol_wage & ~nol_value).sum())
print((nol_value & ~nol_wage).sum())

print("\n", (nol_rel_clau & nol_value).sum(),sep="")
print((nol_rel_clau & ~nol_value).sum())
print((nol_value & ~nol_rel_clau).sum())
```

### Ringkasan Temuan Kolom Value, Wage, dan Release Clause
Ketiga kolom belum bertipe angka, karena nilainya berupa teks dengan simbol € dan akhiran satuan K atau M. Formatnya bervariasi, desimal hanya muncul pada satuan M, dan nilai tanpa akhiran pada Wage tidak selalu nol. Nilai €0 juga tidak selalu sejalan antar kolom, terutama pada Release Clause. Rencana pembersihannya adalah membuang simbol €, mengonversi nilai sesuai satuannya dengan K dikalikan seribu, M dikalikan satu juta, dan nilai tanpa akhiran dibaca apa adanya, membulatkan hasil sebelum dijadikan integer, lalu menyimpannya pada kolom bertipe integer. Perlakuan nilai nol diputuskan berdasarkan perbandingan antar kolom.
